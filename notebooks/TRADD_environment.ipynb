````python
%%writefile download.py
import argparse
import os
#!/usr/bin/env python
""" Downloads FaceForensics public data release
Example usage:
    # source-to-target
    python download.py -d \<'compressed' or 'raw'> \<output folder>
    # self-reenactment
    python download.py -d \<'selfreenactment_compressed' or 'selfreenactment_raw'> \<output folder>
    # cropped self-reenactment images
    python download.py -d selfreenactment_images' \<output folder>
    # only original videos
    python download.py -d original_videos \<output filename>
"""
# -*- coding: utf-8 -*-
import argparse
import os
import urllib
import urllib.request
import tempfile
from os.path import join


SERVER_URL = 'http://kaldir.vc.in.tum.de/FaceForensics/'
TOS_URL = SERVER_URL + 'webpage/FaceForensics_TOS.pdf'
BASE_URL = SERVER_URL + 'v1_cargo/'
ORIGINAL_VIDEOS_URL = BASE_URL + 'original_videos.tar.gz'
RELEASE_DATASET_SIZE = {'raw': '~3.5TB',
                        'compressed': '~130GB',
                        'images': '~135GB'}
DATASET_TYPES = ["raw", "compressed", "selfreenactment_raw",
                 "selfreenactment_compressed", "original_videos",
                 "selfreenactment_images", "source_to_target_images"]
NUM_SAMPLES=5


def get_filelist(filelist_url):
    lines = urllib.request.urlopen(filelist_url)
    video_filenames = []
    for line in lines:
        line = line.decode('utf-8')
        video_filename = line.rstrip('\n')
        video_filenames.append(video_filename)
    return video_filenames


def download_files(filenames, base_url, output_path, sample_only=False):
    os.makedirs(output_path, exist_ok=True)
    num_filenames=len(filenames) if not sample_only else NUM_SAMPLES
    for i, filename in enumerate(filenames):
        if i % 10 == 0:
            print("{}/{}".format(i, num_filenames))
        download_file(base_url + filename, join(output_path, filename))
        if sample_only and i != 0 and i % (NUM_SAMPLES - 1) == 0:
            break
    print("{}/{}".format(num_filenames, num_filenames))


def download_file(url, out_file):
    out_dir = os.path.dirname(out_file)
    if not os.path.isfile(out_file):
        fh, out_file_tmp = tempfile.mkstemp(dir=out_dir)
        f = os.fdopen(fh, 'w')
        f.close()
        urllib.request.urlretrieve(url, out_file_tmp)
        os.rename(out_file_tmp, out_file)
    else:
        print('WARNING: skipping download of existing file ' + out_file)


def main():
    parser = argparse.ArgumentParser(
        description='Downloads FaceForensics public data release.')
    parser.add_argument('output_path', help='directory in which to download')
    parser.add_argument('-d', '--dataset_type', default='compressed',
                        help='Enter which dataset you want to download: '\
                             '"raw", "compressed", "selfreenactment_raw", ' \
                             '"selfreenactment_compressed", ' \
                             '"original_videos" ' \
                             '"source_to_target_images" or ' \
                             '"selfreenactment_images".')
    parser.add_argument('--not_altered', action='store_true',
        help="don't download face2face altered videos")
    parser.add_argument('--not_original', action='store_true',
        help="don't download original videos")
    parser.add_argument('--not_mask', action='store_true',
        help="don't download face2face mask videos")
    parser.add_argument('--not_test', action='store_true',
        help="don't download videos of the test set")
    parser.add_argument('--not_train', action='store_true',
        help="don't download videos of the training set")
    parser.add_argument('--not_val', action='store_true',
        help="don't download videos of the validation set")
    parser.add_argument('--sample_only', action='store_true',
        help='activate this, if you only want to download 5 files per '
             'subfolder')
    args = parser.parse_args()

    # Check for dataset type
    if args.dataset_type not in DATASET_TYPES:
        raise Exception('Wrong dataset type. Please consult "-h" for possible'
                        'options.')

    # TOS
    print('By pressing any key to continue you confirm that you have agreed '\
          'to the FaceForensics terms of use as described at:')
    print(TOS_URL)
    print('***')
    print('Press any key to continue, or CTRL-C to exit.')
    key = input('')

    # Check which videos to download
    downloaded_video_types = []
    if not args.not_altered: downloaded_video_types.append('altered')
    if not args.not_original: downloaded_video_types.append('original')
    if not 'images' in args.dataset_type:
        if not args.not_mask: downloaded_video_types.append('mask')

    # Check which folders to download
    downloaded_folders = []
    if not args.not_test: downloaded_folders.append('test')
    if not args.not_train: downloaded_folders.append('train')
    if not args.not_val: downloaded_folders.append('val')

    # Check for dataset type
    if 'selfreenactment' in args.dataset_type:
        dataset = 'selfreenactment'
        dataset_type = args.dataset_type.replace('selfreenactment_', '')
    else:
        dataset = 'source_to_target'
        dataset_type = args.dataset_type.replace('source_to_target_', '')

    # Warning
    if not args.dataset_type == 'original_videos':
        dataset_filesize = RELEASE_DATASET_SIZE[dataset_type]
        print('***')
        if not args.sample_only:
            print('WARNING: You are downloading the FaceForensics dataset {} of'
                  ' size {}'.format(args.dataset_type, dataset_filesize))
        print(
            'Note that existing scan directories will be skipped. Delete ' \
            'partially downloaded directories to re-download.')
        print('***')
        print('Press any key to continue, or CTRL-C to exit.')
        key = input('')

    # Download
    print('\nDownloading dataset: {}'.format(args.dataset_type))
    if args.dataset_type == 'original_videos':
        print('Please be patient, this may take a while (~2gb)')
        download_file(ORIGINAL_VIDEOS_URL,
                      out_file=join(args.output_path,
                                    'faceforensics_original_videos.tar.gz'))
    else:
        for folder in downloaded_folders:
            if 'images' in args.dataset_type:
                filelist_folder = 'images_' + folder
            else:
                filelist_folder = folder
            filelist_url = BASE_URL + '{}/filelists/{}.txt'.format(dataset,
                                                                filelist_folder)
            filenames = get_filelist(filelist_url)
            print('\nDownloading {}'.format(folder))
            for video_type in downloaded_video_types:
                output_path = join(args.output_path,
                                   'FaceForensics_{}'.format(args.dataset_type),
                                   folder, video_type)
                print('{}/{} > {}'.format(folder, video_type, output_path))
                base_url = BASE_URL + '{}/{}/{}/{}/'.format(dataset,
                                                            dataset_type,
                                                            folder,video_type)
                download_files(filenames, base_url, output_path=output_path,
                               sample_only=args.sample_only)


if __name__ == "__main__":
    main()
````

````python
import cv2
import os
import numpy as np
import pandas as pd
from tqdm.auto import tqdm

# Output directories
OUTPUT_PATH = (
    "/content/drive/MyDrive/TRADD/datasets/processed/frames"
)

REAL_OUTPUT = os.path.join(OUTPUT_PATH, "real")
FAKE_OUTPUT = os.path.join(OUTPUT_PATH, "fake")

os.makedirs(REAL_OUTPUT, exist_ok=True)
os.makedirs(FAKE_OUTPUT, exist_ok=True)

FRAMES_PER_VIDEO = 10

metadata = []

def extract_frames(video_path, output_dir, label, video_name):

    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print("Unable to open:", video_name)
        return

    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    fps = cap.get(cv2.CAP_PROP_FPS)

    if total_frames <= 0:
        cap.release()
        return

    frame_indices = np.linspace(
        0,
        total_frames - 1,
        min(FRAMES_PER_VIDEO, total_frames),
        dtype=int
    )

    video_output = os.path.join(
        output_dir,
        os.path.splitext(video_name)[0]
    )

    os.makedirs(video_output, exist_ok=True)

    for idx, frame_number in enumerate(frame_indices):

        cap.set(cv2.CAP_PROP_POS_FRAMES, int(frame_number))

        ret, frame = cap.read()

        if not ret:
            continue

        frame_name = f"frame_{idx:03d}.jpg"

        frame_path = os.path.join(
            video_output,
            frame_name
        )

        success = cv2.imwrite(
            frame_path,
            frame,
            [cv2.IMWRITE_JPEG_QUALITY, 95]
        )

        if success:
            metadata.append({
                "video_name": video_name,
                "label": label,
                "class": "REAL" if label == 0 else "FAKE",
                "frame_number": int(frame_number),
                "fps": fps,
                "total_frames": total_frames,
                "frame_path": frame_path
            })

    cap.release()


# Extract original videos
print("Extracting REAL videos...")

for video in tqdm(real_videos):
    video_path = os.path.join(REAL_PATH, video)

    extract_frames(
        video_path,
        REAL_OUTPUT,
        0,
        video
    )


# Extract manipulated videos
print("\nExtracting FAKE videos...")

for video in tqdm(fake_videos):
    video_path = os.path.join(FAKE_PATH, video)

    extract_frames(
        video_path,
        FAKE_OUTPUT,
        1,
        video
    )


# Save metadata
df = pd.DataFrame(metadata)

metadata_path = os.path.join(
    OUTPUT_PATH,
    "frame_metadata.csv"
)

df.to_csv(metadata_path, index=False)

print("\nFRAME EXTRACTION COMPLETED")
print("Total extracted frames:", len(df))
print("REAL frames:", (df["label"] == 0).sum())
print("FAKE frames:", (df["label"] == 1).sum())
print("Metadata saved:", metadata_path)
````

````python
import os
import cv2
import pandas as pd
import numpy as np
from tqdm.notebook import tqdm

FACE_SIZE = 224
PADDING = 0.25

crop_records = []
failed_frames = []

for _, row in tqdm(df.iterrows(), total=len(df)):

    image = cv2.imread(row["frame_path"])

    if image is None:
        failed_frames.append(row["frame_path"])
        continue

    height, width = image.shape[:2]

    detector.setInputSize((width, height))
    _, faces = detector.detect(image)

    if faces is None or len(faces) == 0:
        failed_frames.append(row["frame_path"])
        continue

    # Select the largest detected face
    largest_face = max(
        faces,
        key=lambda face: face[2] * face[3]
    )

    x, y, w, h = largest_face[:4]
    confidence = float(largest_face[-1])

    # Make a square crop with padding
    cx = x + w / 2
    cy = y + h / 2

    side = max(w, h) * (1 + PADDING)

    x1 = max(0, int(cx - side / 2))
    y1 = max(0, int(cy - side / 2))
    x2 = min(width, int(cx + side / 2))
    y2 = min(height, int(cy + side / 2))

    face_crop = image[y1:y2, x1:x2]

    if face_crop.size == 0:
        failed_frames.append(row["frame_path"])
        continue

    face_crop = cv2.resize(
        face_crop,
        (FACE_SIZE, FACE_SIZE),
        interpolation=cv2.INTER_AREA
    )

    video_id = os.path.splitext(row["video_name"])[0]

    save_dir = os.path.join(
        OUTPUT_PATH,
        row["class"],
        video_id
    )

    os.makedirs(save_dir, exist_ok=True)

    save_path = os.path.join(
        save_dir,
        os.path.basename(row["frame_path"])
    )

    cv2.imwrite(save_path, face_crop)

    crop_records.append({
        **row.to_dict(),
        "face_crop_path": save_path,
        "face_confidence": confidence,
        "face_x": int(x),
        "face_y": int(y),
        "face_width": int(w),
        "face_height": int(h)
    })

# Save metadata
crop_df = pd.DataFrame(crop_records)

crop_metadata_path = os.path.join(
    OUTPUT_PATH,
    "face_metadata.csv"
)

crop_df.to_csv(crop_metadata_path, index=False)

print("\nFACE EXTRACTION COMPLETED")
print("Successfully cropped:", len(crop_records))
print("Failed frames:", len(failed_frames))
print("\nClass distribution:")
print(crop_df["class"].value_counts())
print("\nMetadata saved:", crop_metadata_path)
````

````python
import os
import random
import pandas as pd

# Load face metadata
metadata_path = (
    "/content/drive/MyDrive/TRADD/datasets/processed/"
    "faces/face_metadata.csv"
)

df = pd.read_csv(metadata_path)

# Five connected identity groups
identity_groups = [
    ["033", "097"],
    ["183", "253"],
    ["210", "241"],
    ["252", "266"],
    ["339", "392"]
]

# Assign groups to splits reproducibly
groups = identity_groups.copy()
random.Random(42).shuffle(groups)

split_assignment = {}

for i, group in enumerate(groups):
    if i < 3:
        split = "train"
    elif i == 3:
        split = "validation"
    else:
        split = "test"

    for identity in group:
        split_assignment[identity] = split

# Assign each video to its identity group
def get_video_split(video_name):
    video_id = os.path.splitext(video_name)[0]

    # Original video
    if "_" not in video_id:
        identity = video_id
        return split_assignment[identity]

    # Manipulated video: both source identities
    source_ids = video_id.split("_")

    splits = {
        split_assignment[identity]
        for identity in source_ids
    }

    if len(splits) != 1:
        raise ValueError(
            f"Identity leakage detected in {video_name}"
        )

    return splits.pop()

df["split"] = df["video_name"].apply(get_video_split)

# Save split metadata
output_path = (
    "/content/drive/MyDrive/TRADD/datasets/processed/"
    "faces/face_metadata_split.csv"
)

df.to_csv(output_path, index=False)

# Display distribution
print("SPLIT DISTRIBUTION")
print(
    df.groupby(["split", "class"])
      .size()
      .unstack(fill_value=0)
)

print("\nVIDEO DISTRIBUTION")
print(
    df.groupby("split")["video_name"]
      .nunique()
)

print("\nIdentity group assignments:")
for group in groups:
    print(group, "->", split_assignment[group[0]])

print("\nSaved:", output_path)
````

````python
class TRADD_Dataset(Dataset):

    def __init__(self, dataframe, transform=None):

        self.data = dataframe.reset_index(drop=True)
        self.transform = transform

        self.label_mapping = {
            "REAL": 0,
            "FAKE": 1
        }

    def __len__(self):
        return len(self.data)

    def __getitem__(self, index):

        row = self.data.iloc[index]

        image_path = row["face_crop_path"]
        label = self.label_mapping[row["class"]]

        image = Image.open(image_path).convert("RGB")

        if self.transform:
            image = self.transform(image)

        return image, torch.tensor(label, dtype=torch.long)
````

````python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

def calculate_metrics(labels, predictions, probabilities):

    return {
        "accuracy": accuracy_score(labels, predictions),
        "precision": precision_score(
            labels, predictions, zero_division=0
        ),
        "recall": recall_score(
            labels, predictions, zero_division=0
        ),
        "f1_score": f1_score(
            labels, predictions, zero_division=0
        ),
        "roc_auc": roc_auc_score(
            labels, probabilities
        )
    }

print("Evaluation metrics configured!")
````

````python
def train_one_epoch(model, loader, criterion, optimizer, device):

    model.train()

    total_loss = 0
    all_labels = []
    all_predictions = []

    for images, labels in loader:

        images = images.to(device)
        labels = labels.to(device)

        # Clear gradients from previous batch
        optimizer.zero_grad()

        # Forward propagation
        outputs = model(images)

        # Calculate loss
        loss = criterion(outputs, labels)

        # Backpropagation
        loss.backward()

        # Update model weights
        optimizer.step()

        total_loss += loss.item() * images.size(0)

        predictions = outputs.argmax(dim=1)

        all_labels.extend(labels.cpu().numpy())
        all_predictions.extend(predictions.cpu().numpy())

    epoch_loss = total_loss / len(loader.dataset)

    accuracy = accuracy_score(
        all_labels,
        all_predictions
    )

    return epoch_loss, accuracy


def validate_one_epoch(model, loader, criterion, device):

    model.eval()

    total_loss = 0
    all_labels = []
    all_predictions = []
    all_probabilities = []

    with torch.no_grad():

        for images, labels in loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)

            loss = criterion(outputs, labels)

            total_loss += loss.item() * images.size(0)

            probabilities = torch.softmax(outputs, dim=1)

            predictions = outputs.argmax(dim=1)

            all_labels.extend(labels.cpu().numpy())
            all_predictions.extend(predictions.cpu().numpy())
            all_probabilities.extend(
                probabilities[:, 1].cpu().numpy()
            )

    epoch_loss = total_loss / len(loader.dataset)

    metrics = calculate_metrics(
        all_labels,
        all_predictions,
        all_probabilities
    )

    return epoch_loss, metrics
````

````python
import copy
import os

EPOCHS = 15
PATIENCE = 4

best_val_loss = float("inf")
epochs_without_improvement = 0

best_model_state = None

history = []

MODEL_SAVE_PATH = os.path.join(
    PROJECT_PATH,
    "models",
    "mobilenetv3_baseline_best.pth"
)

for epoch in range(EPOCHS):

    train_loss, train_accuracy = train_one_epoch(
        model,
        train_loader,
        criterion,
        optimizer,
        device
    )

    val_loss, val_metrics = validate_one_epoch(
        model,
        val_loader,
        criterion,
        device
    )

    scheduler.step(val_loss)

    current_lr = optimizer.param_groups[0]["lr"]

    history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_accuracy": train_accuracy,
        "val_loss": val_loss,
        "val_accuracy": val_metrics["accuracy"],
        "val_f1": val_metrics["f1_score"],
        "val_auc": val_metrics["roc_auc"]
    })

    print(f"\nEpoch {epoch + 1}/{EPOCHS}")
    print(f"Train Loss: {train_loss:.4f}")
    print(f"Train Accuracy: {train_accuracy:.4f}")
    print(f"Validation Loss: {val_loss:.4f}")
    print(f"Validation Accuracy: {val_metrics['accuracy']:.4f}")
    print(f"Validation F1: {val_metrics['f1_score']:.4f}")
    print(f"Validation AUC: {val_metrics['roc_auc']:.4f}")
    print(f"Learning Rate: {current_lr:.6f}")

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        best_model_state = copy.deepcopy(model.state_dict())

        torch.save(
            best_model_state,
            MODEL_SAVE_PATH
        )

        epochs_without_improvement = 0

        print("Best model saved!")

    else:

        epochs_without_improvement += 1

    if epochs_without_improvement >= PATIENCE:

        print("\nEarly stopping triggered.")
        break

print("\nTraining completed.")
````

````python
def run_experiment(
    model,
    optimizer,
    scheduler,
    experiment_name,
    epochs=15,
    patience=4
):

    best_loss = float("inf")
    best_state = None
    wait = 0
    experiment_history = []

    for epoch in range(epochs):

        train_loss, train_acc = train_one_epoch(
            model,
            train_loader,
            criterion,
            optimizer,
            device
        )

        val_loss, val_metrics = validate_one_epoch(
            model,
            val_loader,
            criterion,
            device
        )

        scheduler.step(val_loss)

        experiment_history.append({
            "epoch": epoch + 1,
            "train_loss": train_loss,
            "train_accuracy": train_acc,
            "val_loss": val_loss,
            "val_accuracy": val_metrics["accuracy"],
            "val_f1": val_metrics["f1_score"],
            "val_auc": val_metrics["roc_auc"]
        })

        print(
            f"Epoch {epoch+1}/{epochs} | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {val_loss:.4f} | "
            f"Val Accuracy: {val_metrics['accuracy']:.4f}"
        )

        if val_loss < best_loss:

            best_loss = val_loss
            best_state = copy.deepcopy(model.state_dict())
            wait = 0

        else:
            wait += 1

        if wait >= patience:
            print("Early stopping.")
            break

    model.load_state_dict(best_state)

    return model, experiment_history
````

````python
import os
import glob
import random
import pandas as pd
from collections import defaultdict

BASE = "/content/drive/MyDrive/TRADD/datasets/raw/FaceForensicsPP_expansion"

REAL_DIR = os.path.join(
    BASE, "original_sequences/youtube/c23/videos"
)

METHODS = ["Deepfakes", "Face2Face", "FaceSwap", "NeuralTextures"]

# 1. Rebuild identity graph
real_files = glob.glob(os.path.join(REAL_DIR, "*.mp4"))

real_ids = {
    os.path.splitext(os.path.basename(f))[0]
    for f in real_files
}

graph = defaultdict(set)

for vid in real_ids:
    graph[vid]

for method in METHODS:
    fake_dir = os.path.join(
        BASE, "manipulated_sequences", method, "c23", "videos"
    )

    for path in glob.glob(os.path.join(fake_dir, "*.mp4")):
        name = os.path.splitext(os.path.basename(path))[0]
        target, source = name.split("_")

        graph[target].add(source)
        graph[source].add(target)

# 2. Find connected components
visited = set()
groups = []

for node in graph:
    if node in visited:
        continue

    stack = [node]
    group = []

    while stack:
        current = stack.pop()

        if current in visited:
            continue

        visited.add(current)
        group.append(current)

        for neighbour in graph[current]:
            if neighbour not in visited:
                stack.append(neighbour)

    groups.append(sorted(group))

# 3. Shuffle and assign groups
random.seed(42)
random.shuffle(groups)

train_groups = groups[:30]
val_groups = groups[30:40]
test_groups = groups[40:50]

group_to_split = {}

for split, selected_groups in [
    ("train", train_groups),
    ("val", val_groups),
    ("test", test_groups)
]:
    for group in selected_groups:
        for identity in group:
            group_to_split[identity] = split

# 4. Build video manifest
records = []

# Original videos
for path in real_files:
    vid = os.path.splitext(os.path.basename(path))[0]

    records.append({
        "video_name": vid + ".mp4",
        "video_path": path,
        "label": 0,
        "class": "REAL",
        "manipulation": "Original",
        "identity_group": next(
            i for i, g in enumerate(groups) if vid in g
        ),
        "split": group_to_split[vid]
    })

# Manipulated videos
for method in METHODS:
    fake_dir = os.path.join(
        BASE, "manipulated_sequences", method, "c23", "videos"
    )

    for path in glob.glob(os.path.join(fake_dir, "*.mp4")):
        name = os.path.splitext(os.path.basename(path))[0]
        target, source = name.split("_")

        assert group_to_split[target] == group_to_split[source]

        records.append({
            "video_name": name + ".mp4",
            "video_path": path,
            "label": 1,
            "class": "FAKE",
            "manipulation": method,
            "identity_group": next(
                i for i, g in enumerate(groups) if target in g
            ),
            "split": group_to_split[target]
        })

# 5. Save CSV
manifest = pd.DataFrame(records)

OUTPUT = os.path.join(BASE, "video_split_manifest.csv")
manifest.to_csv(OUTPUT, index=False)

# 6. Verify results
print("Total videos:", len(manifest))
print("\nSplit distribution:")
print(pd.crosstab(manifest["split"], manifest["class"]))

print("\nManipulation distribution:")
print(pd.crosstab(manifest["split"], manifest["manipulation"]))

print("\nSaved manifest:", OUTPUT)

# Verify no identity group crosses splits
assert manifest.groupby("identity_group")["split"].nunique().max() == 1

print("\nIdentity leakage check: PASSED")
````

````python

import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import models

# Create an orthonormal 8x8 DCT transformation matrix
def create_dct_matrix(N=8):
    matrix = torch.zeros(N, N)

    for k in range(N):
        alpha = (1 / N) ** 0.5 if k == 0 else (2 / N) ** 0.5

        for n in range(N):
            matrix[k, n] = alpha * torch.cos(
                torch.tensor(
                    torch.pi * (2 * n + 1) * k / (2 * N)
                )
            )

    return matrix


class TRADD(nn.Module):

    def __init__(self, num_classes=2):

        super().__init__()

        # 1. Spatial branch: pretrained MobileNetV3 Small
        backbone = models.mobilenet_v3_small(
            weights=models.MobileNet_V3_Small_Weights.DEFAULT
        )

        self.spatial_features = backbone.features
        self.spatial_pool = backbone.avgpool

        # 2. ImageNet normalization values
        self.register_buffer(
            "mean",
            torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1)
        )

        self.register_buffer(
            "std",
            torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1)
        )

        # 3. DCT transformation matrix
        self.register_buffer("dct_matrix", create_dct_matrix())

        # 4. Frequency branch
        self.frequency_branch = nn.Sequential(
            nn.Conv2d(64, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),

            nn.Conv2d(32, 64, kernel_size=3, stride=2, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),

            nn.Conv2d(64, 128, kernel_size=3, stride=2, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),

            nn.AdaptiveAvgPool2d(1)
        )

        # 5. Feature fusion and classification
        self.classifier = nn.Sequential(
            nn.Linear(576 + 128, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(256, num_classes)
        )

    def extract_frequency(self, x):

        # Reverse ImageNet normalization
        x = (x * self.std + self.mean).clamp(0, 1)

        # Convert RGB image to grayscale
        gray = (
            0.299 * x[:, 0:1] +
            0.587 * x[:, 1:2] +
            0.114 * x[:, 2:3]
        )

        # Divide image into non-overlapping 8x8 blocks
        blocks = F.unfold(
            gray,
            kernel_size=8,
            stride=8
        )

        batch_size = x.shape[0]

        blocks = blocks.transpose(1, 2).reshape(
            batch_size, -1, 8, 8
        )

        # Apply 2D DCT to every block
        dct = torch.einsum(
            "ij,bnjk,lk->bnil",
            self.dct_matrix,
            blocks,
            self.dct_matrix
        )

        # Arrange coefficients as 64 feature maps
        frequency = dct.reshape(
            batch_size, -1, 64
        ).transpose(1, 2).reshape(
            batch_size, 64, 28, 28
        )

        # Compress the coefficient range while preserving signs
        frequency = (
            torch.sign(frequency) *
            torch.log1p(torch.abs(frequency))
        )

        return frequency

    def forward(self, x):

        # Spatial features
        spatial = self.spatial_features(x)
        spatial = self.spatial_pool(spatial)
        spatial = torch.flatten(spatial, 1)

        # Frequency features
        frequency = self.extract_frequency(x)
        frequency = self.frequency_branch(frequency)
        frequency = torch.flatten(frequency, 1)

        # Combine both feature representations
        combined = torch.cat(
            [spatial, frequency], dim=1
        )

        # REAL / FAKE prediction
        output = self.classifier(combined)

        return output


# Initialize model
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = TRADD().to(device)

print("Model initialized successfully!")
print("Device:", device)
````

````python

import os
import pandas as pd
import torch
from PIL import Image
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

ROOT = "/content/drive/MyDrive/TRADD"
FACE_DIR = ROOT + "/datasets/processed/faces_expansion"

metadata = pd.read_csv(
    FACE_DIR + "/face_metadata.csv"
)

# Image transformations
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ColorJitter(
        brightness=0.1,
        contrast=0.1,
        saturation=0.1
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])

eval_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])


class TRADD_Dataset(Dataset):

    def __init__(self, dataframe, transform=None):
        self.data = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.data)

    def __getitem__(self, index):
        row = self.data.iloc[index]

        image = Image.open(row["face_path"]).convert("RGB")
        label = int(row["label"])

        if self.transform:
            image = self.transform(image)

        return image, label


# Separate datasets using existing split column
train_data = metadata[metadata["split"] == "train"]
val_data = metadata[metadata["split"] == "val"]
test_data = metadata[metadata["split"] == "test"]

train_dataset = TRADD_Dataset(train_data, train_transform)
val_dataset = TRADD_Dataset(val_data, eval_transform)
test_dataset = TRADD_Dataset(test_data, eval_transform)

# Create DataLoaders
train_loader = DataLoader(
    train_dataset, batch_size=32,
    shuffle=True, num_workers=2
)

val_loader = DataLoader(
    val_dataset, batch_size=32,
    shuffle=False, num_workers=2
)

test_loader = DataLoader(
    test_dataset, batch_size=32,
    shuffle=False, num_workers=2
)

print("Training frames:", len(train_dataset))
print("Validation frames:", len(val_dataset))
print("Testing frames:", len(test_dataset))
print("DataLoaders recreated successfully!")
````

````python

import os
import copy
import torch
import torch.nn as nn
import numpy as np

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)

# Training configuration
EPOCHS = 15
PATIENCE = 4
LEARNING_RATE = 0.0001

# Class weights: REAL=0, FAKE=1
class_counts = torch.tensor([600, 2400], dtype=torch.float32)

class_weights = class_counts.sum() / (2 * class_counts)
class_weights = class_weights.to(device)

criterion = nn.CrossEntropyLoss(weight=class_weights)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=0.0001
)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=2
)

SAVE_DIR = "/content/drive/MyDrive/TRADD/models"
os.makedirs(SAVE_DIR, exist_ok=True)

BEST_PATH = os.path.join(SAVE_DIR, "tradd_spatial_frequency_best.pth")


# One training epoch
def train_one_epoch(model, loader):

    model.train()
    total_loss = 0

    for images, labels in loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        total_loss += loss.item() * images.size(0)

    return total_loss / len(loader.dataset)


# Validation
def evaluate(model, loader):

    model.eval()

    total_loss = 0
    all_labels = []
    all_predictions = []
    all_probabilities = []

    with torch.no_grad():

        for images, labels in loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            probabilities = torch.softmax(outputs, dim=1)
            predictions = outputs.argmax(dim=1)

            total_loss += loss.item() * images.size(0)

            all_labels.extend(labels.cpu().numpy())
            all_predictions.extend(predictions.cpu().numpy())
            all_probabilities.extend(
                probabilities[:, 1].cpu().numpy()
            )

    labels = np.array(all_labels)
    predictions = np.array(all_predictions)
    probabilities = np.array(all_probabilities)

    metrics = {
        "loss": total_loss / len(loader.dataset),
        "accuracy": accuracy_score(labels, predictions),
        "precision": precision_score(
            labels, predictions, zero_division=0
        ),
        "recall": recall_score(
            labels, predictions, zero_division=0
        ),
        "f1": f1_score(
            labels, predictions, zero_division=0
        ),
        "auc": roc_auc_score(labels, probabilities)
    }

    return metrics


# Training loop with early stopping
best_val_loss = float("inf")
best_epoch = 0
patience_counter = 0
history = []

for epoch in range(EPOCHS):

    train_loss = train_one_epoch(model, train_loader)
    val_metrics = evaluate(model, val_loader)

    scheduler.step(val_metrics["loss"])

    current_lr = optimizer.param_groups[0]["lr"]

    history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        **val_metrics
    })

    print(
        f"Epoch [{epoch+1}/{EPOCHS}] "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_metrics['loss']:.4f} | "
        f"Val Acc: {val_metrics['accuracy']:.4f} | "
        f"Val F1: {val_metrics['f1']:.4f} | "
        f"Val AUC: {val_metrics['auc']:.4f} | "
        f"LR: {current_lr:.6f}"
    )

    # Save best model based on validation loss
    if val_metrics["loss"] < best_val_loss:

        best_val_loss = val_metrics["loss"]
        best_epoch = epoch + 1
        patience_counter = 0

        torch.save({
            "epoch": best_epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "val_loss": best_val_loss
        }, BEST_PATH)

        print("  Best checkpoint saved.")

    else:
        patience_counter += 1

    if patience_counter >= PATIENCE:
        print("\nEarly stopping triggered.")
        break


print("\nTraining completed.")
print("Best epoch:", best_epoch)
print("Best validation loss:", round(best_val_loss, 4))
print("Checkpoint:", BEST_PATH)
````

````python

class TRADD(nn.Module):

    def __init__(self, num_classes=2, use_dct=True):

        super().__init__()

        self.use_dct = use_dct

        # Spatial branch
        backbone = models.mobilenet_v3_small(
            weights=models.MobileNet_V3_Small_Weights.DEFAULT
        )

        self.spatial_features = backbone.features
        self.spatial_pool = backbone.avgpool

        # ImageNet normalization
        self.register_buffer(
            "mean",
            torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1)
        )

        self.register_buffer(
            "std",
            torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1)
        )

        # DCT matrix
        self.register_buffer("dct_matrix", create_dct_matrix())

        # Frequency branch
        self.frequency_branch = nn.Sequential(
            nn.Conv2d(64, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),

            nn.Conv2d(32, 64, 3, stride=2, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),

            nn.Conv2d(64, 128, 3, stride=2, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),

            nn.AdaptiveAvgPool2d(1)
        )

        # Identical classifier for both configurations
        self.classifier = nn.Sequential(
            nn.Linear(576 + 128, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(256, num_classes)
        )

    def extract_frequency(self, x):

        x = (x * self.std + self.mean).clamp(0, 1)

        gray = (
            0.299 * x[:, 0:1] +
            0.587 * x[:, 1:2] +
            0.114 * x[:, 2:3]
        )

        blocks = F.unfold(
            gray, kernel_size=8, stride=8
        )

        batch_size = x.shape[0]

        blocks = blocks.transpose(1, 2).reshape(
            batch_size, -1, 8, 8
        )

        dct = torch.einsum(
            "ij,bnjk,lk->bnil",
            self.dct_matrix,
            blocks,
            self.dct_matrix
        )

        frequency = dct.reshape(
            batch_size, -1, 64
        ).transpose(1, 2).reshape(
            batch_size, 64, 28, 28
        )

        frequency = (
            torch.sign(frequency) *
            torch.log1p(torch.abs(frequency))
        )

        return frequency

    def forward(self, x):

        # Spatial features
        spatial = self.spatial_features(x)
        spatial = self.spatial_pool(spatial)
        spatial = torch.flatten(spatial, 1)

        # Frequency features or zero vector
        if self.use_dct:

            frequency = self.extract_frequency(x)
            frequency = self.frequency_branch(frequency)
            frequency = torch.flatten(frequency, 1)

        else:

            frequency = torch.zeros(
                x.size(0), 128,
                device=x.device,
                dtype=spatial.dtype
            )

        # Feature fusion
        combined = torch.cat(
            [spatial, frequency], dim=1
        )

        return self.classifier(combined)
````

````python

import os
import json
import subprocess
import pandas as pd
from pathlib import Path

VIDEO_DIR = ROOT + "/datasets/raw/FaceForensicsPP_expansion"
OUTPUT_CSV = ROOT + "/datasets/processed/video_encoding_audit.csv"

# Find all source videos
video_paths = list(Path(VIDEO_DIR).rglob("*.mp4"))

print("Videos found:", len(video_paths))

records = []

for i, path in enumerate(video_paths):
    command = [
        "ffprobe", "-v", "error",
        "-select_streams", "v:0",
        "-show_entries",
        "stream=codec_name,width,height,pix_fmt,r_frame_rate,bit_rate",
        "-show_entries",
        "format=duration,size,bit_rate",
        "-of", "json",
        str(path)
    ]

    try:
        result = subprocess.run(
            command,
            capture_output=True,
            text=True,
            timeout=20
        )

        info = json.loads(result.stdout)
        stream = info.get("streams", [{}])[0]
        fmt = info.get("format", {})

        parts = path.parts
        category = next(
            (x for x in parts if x in [
                "original_sequences",
                "Deepfakes",
                "Face2Face",
                "FaceSwap",
                "NeuralTextures"
            ]),
            "Unknown"
        )

        if category == "original_sequences":
            category = "Original"

        records.append({
            "video": path.name,
            "category": category,
            "codec": stream.get("codec_name"),
            "width": stream.get("width"),
            "height": stream.get("height"),
            "pixel_format": stream.get("pix_fmt"),
            "fps": stream.get("r_frame_rate"),
            "duration": float(fmt.get("duration", 0) or 0),
            "file_size_MB": round(
                int(fmt.get("size", 0) or 0) / (1024**2), 3
            ),
            "bitrate_kbps": round(
                int(stream.get("bit_rate") or fmt.get("bit_rate") or 0)
                / 1000, 2
            ),
            "error": None
        })

    except Exception as e:
        records.append({
            "video": path.name,
            "category": category,
            "error": str(e)
        })

    if (i + 1) % 50 == 0:
        print(f"Processed {i+1}/{len(video_paths)} videos")

audit = pd.DataFrame(records)
audit.to_csv(OUTPUT_CSV, index=False)

print("\nSaved audit:", OUTPUT_CSV)
print("Successful:", audit["error"].isna().sum())
print("Failed:", audit["error"].notna().sum())

# Summarise technical properties by category
for column in ["codec", "width", "height", "pixel_format", "fps"]:
    print(f"\n{column.upper()} DISTRIBUTION")
    print(pd.crosstab(audit["category"], audit[column]))

print("\nNUMERIC STATISTICS BY CATEGORY")
print(
    audit.groupby("category")[
        ["duration", "file_size_MB", "bitrate_kbps"]
    ].agg(["count", "mean", "median", "std"]).round(2)
)
````

````python

import os
import numpy as np
import pandas as pd

audit_path = ROOT + "/datasets/processed/video_encoding_audit.csv"
audit = pd.read_csv(audit_path)

# Create a lookup table for original videos
originals = audit[audit["category"] == "Original"].copy()

originals["video_id"] = originals["video"].str.replace(
    ".mp4", "", regex=False
)

original_lookup = originals.set_index("video_id").to_dict("index")

# Compare each manipulated video with its two original IDs
comparisons = []

fake_data = audit[audit["category"] != "Original"]

for _, row in fake_data.iterrows():
    ids = row["video"].replace(".mp4", "").split("_")

    if len(ids) != 2:
        continue

    original_a = original_lookup.get(ids[0])
    original_b = original_lookup.get(ids[1])

    for position, original in [("ID_A", original_a), ("ID_B", original_b)]:
        if original is None:
            continue

        comparisons.append({
            "category": row["category"],
            "video": row["video"],
            "original_id": position,
            "width_difference": abs(
                row["width"] - original["width"]
            ),
            "height_difference": abs(
                row["height"] - original["height"]
            ),
            "duration_difference": abs(
                row["duration"] - original["duration"]
            ),
            "bitrate_difference": abs(
                row["bitrate_kbps"] - original["bitrate_kbps"]
            )
        })

comparison_df = pd.DataFrame(comparisons)

print("COMPARISON RECORDS:", len(comparison_df))

print("\nMEDIAN DIFFERENCES BY MANIPULATION")
print(
    comparison_df.groupby(["category", "original_id"])[
        ["width_difference", "height_difference",
         "duration_difference", "bitrate_difference"]
    ].median().round(2)
)

print("\nEXACT RESOLUTION MATCH RATE (%)")
comparison_df["resolution_match"] = (
    (comparison_df["width_difference"] == 0) &
    (comparison_df["height_difference"] == 0)
)

print(
    comparison_df.groupby(["category", "original_id"])[
        "resolution_match"
    ].mean().mul(100).round(1)
)

print("\nEXACT DURATION MATCH RATE (%)")
comparison_df["duration_match"] = (
    comparison_df["duration_difference"] < 0.05
)

print(
    comparison_df.groupby(["category", "original_id"])[
        "duration_match"
    ].mean().mul(100).round(1)
)
````

````python

import torch
import torch.nn as nn
import pandas as pd
from sklearn.metrics import roc_auc_score

model = TRADD(use_dct=False).to(device)

criterion = nn.CrossEntropyLoss(
    weight=torch.tensor([2.5, 0.625], device=device)
)

optimizer = torch.optim.AdamW(
    model.parameters(), lr=1e-4, weight_decay=1e-4
)

best_auc = 0
patience = 0
history = []

for epoch in range(15):
    model.train()
    total_loss = 0

    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        loss = criterion(model(images), labels)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    model.eval()
    predictions = []

    with torch.no_grad():
        for images, labels in val_loader:
            probabilities = torch.softmax(
                model(images.to(device)), dim=1
            )[:, 1]
            predictions.extend(probabilities.cpu().numpy())

    val = val_loader.dataset.data.copy()
    val["probability"] = predictions

    videos = val.groupby(
        ["manipulation", "video_name"], as_index=False
    ).agg(label=("label", "first"),
          probability=("probability", "mean"))

    auc = roc_auc_score(videos["label"], videos["probability"])
    train_loss = total_loss / len(train_loader)

    history.append([epoch + 1, train_loss, auc])

    print(f"Epoch {epoch+1}: Loss={train_loss:.4f}, Video AUC={auc:.4f}")

    if auc > best_auc:
        best_auc = auc
        patience = 0
        torch.save(model.state_dict(),
                   "/content/drive/MyDrive/TRADD/models/rgb_best.pth")
    else:
        patience += 1

    if patience == 4:
        break

print("Best validation video AUC:", best_auc)
````

````python

from sklearn.metrics import (
    balanced_accuracy_score, accuracy_score,
    precision_score, recall_score, f1_score,
    confusion_matrix
)
import numpy as np

def evaluate_model(model):
    def get_predictions(loader):
        model.eval()
        probs = []

        with torch.no_grad():
            for images, labels in loader:
                output = model(images.to(device))
                probs.extend(
                    torch.softmax(output, dim=1)[:, 1].cpu().numpy()
                )

        return np.array(probs)

    val_probs = get_predictions(val_loader)
    test_probs = get_predictions(test_loader)

    val_meta = val_loader.dataset.data
    test_meta = test_loader.dataset.data

    def aggregate(meta, probs):
        df = meta.copy()
        df["probability"] = probs
        return df.groupby(
            ["manipulation", "video_name"], as_index=False
        ).agg(label=("label", "first"),
              probability=("probability", "mean"))

    val = aggregate(val_meta, val_probs)
    test = aggregate(test_meta, test_probs)

    thresholds = np.arange(0.05, 0.96, 0.01)
    threshold = max(
        thresholds,
        key=lambda t: balanced_accuracy_score(
            val["label"], (val["probability"] >= t).astype(int)
        )
    )

    y_true = test["label"]
    y_pred = (test["probability"] >= threshold).astype(int)

    print("Threshold:", round(threshold, 2))
    print("Test AUC:", round(roc_auc_score(y_true, test["probability"]), 4))
    print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
    print("Balanced Accuracy:", round(balanced_accuracy_score(y_true, y_pred), 4))
    print("Precision:", round(precision_score(y_true, y_pred), 4))
    print("Recall:", round(recall_score(y_true, y_pred), 4))
    print("Macro F1:", round(f1_score(y_true, y_pred, average="macro"), 4))
    print("Confusion Matrix:\n", confusion_matrix(y_true, y_pred))

print("RGB-ONLY RESULTS")
evaluate_model(rgb_model)

print("\nRGB + DCT RESULTS")
evaluate_model(hybrid_model)
````

````python

from torch.utils.data import Dataset, DataLoader
from PIL import Image
import torch
import pandas as pd

class VideoSequenceDataset(Dataset):
    def __init__(self, data, transform):
        self.data = data
        self.transform = transform

        self.groups = [
            group.sort_values("frame_number")
            for _, group in data.groupby(
                ["manipulation", "video_name"]
            )
        ]

    def __len__(self):
        return len(self.groups)

    def __getitem__(self, idx):
        group = self.groups[idx]

        frames = []
        for path in group["face_path"]:
            image = Image.open(path).convert("RGB")
            frames.append(self.transform(image))

        return torch.stack(frames), int(group["label"].iloc[0])


transform = val_loader.dataset.transform

train_sequences = VideoSequenceDataset(train_data, transform)
val_sequences = VideoSequenceDataset(val_data, transform)
test_sequences = VideoSequenceDataset(test_data, transform)

train_video_loader = DataLoader(train_sequences, batch_size=4, shuffle=True)
val_video_loader = DataLoader(val_sequences, batch_size=4, shuffle=False)
test_video_loader = DataLoader(test_sequences, batch_size=4, shuffle=False)

print("Training videos:", len(train_sequences))
print("Validation videos:", len(val_sequences))
print("Test videos:", len(test_sequences))

frames, labels = next(iter(train_video_loader))
print("Sequence shape:", frames.shape)
````

````python
def extract_video_features(loader, model):
    model.eval()
    all_features, all_labels = [], []

    with torch.no_grad():
        for frames, labels in loader:
            b, t, c, h, w = frames.shape
            frames = frames.reshape(b * t, c, h, w).to(device)

            spatial = model.spatial_pool(
                model.spatial_features(frames)
            ).flatten(1)

            frequency = model.frequency_branch(
                model.extract_frequency(frames)
            ).flatten(1)

            fused = torch.cat([spatial, frequency], dim=1)
            all_features.append(fused.reshape(b, t, 704).cpu())
            all_labels.append(labels)

    return torch.cat(all_features), torch.cat(all_labels)


train_features, train_labels = extract_video_features(
    train_video_loader, hybrid_model
)

val_features, val_labels = extract_video_features(
    val_video_loader, hybrid_model
)

test_features, test_labels = extract_video_features(
    test_video_loader, hybrid_model
)

print("Training features:", train_features.shape)
print("Validation features:", val_features.shape)
print("Test features:", test_features.shape)
````

````python
import torch.nn as nn

class TemporalTRADD(nn.Module):
    def __init__(self, input_size=704, hidden_size=128):
        super().__init__()

        self.gru = nn.GRU(
            input_size=input_size,
            hidden_size=hidden_size,
            batch_first=True,
            num_layers=1
        )

        self.classifier = nn.Sequential(
            nn.Linear(hidden_size, 64),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(64, 2)
        )

    def forward(self, x):
        _, hidden = self.gru(x)
        return self.classifier(hidden[-1])


temporal_model = TemporalTRADD().to(device)

print(temporal_model)
````

````python

def get_probabilities(loader):
    temporal_model.eval()
    probs, labels = [], []

    with torch.no_grad():
        for x, y in loader:
            output = temporal_model(x.to(device))
            probs.extend(torch.softmax(output, dim=1)[:, 1].cpu().numpy())
            labels.extend(y.numpy())

    return np.array(probs), np.array(labels)


val_probs, val_y = get_probabilities(val_feature_loader)
test_probs, test_y = get_probabilities(test_feature_loader)

# Confidence of the predicted class
val_confidence = np.maximum(val_probs, 1 - val_probs)
val_predictions = (val_probs >= 0.5).astype(int)

# Select the lowest confidence threshold giving <=10% validation error
threshold = 1.01

for t in np.sort(np.unique(val_confidence)):
    accepted = val_confidence >= t

    if accepted.sum() > 0:
        error = np.mean(val_predictions[accepted] != val_y[accepted])

        if error <= 0.10:
            threshold = t
            break

print("Selected confidence threshold:", round(threshold, 4))
````

````python

def find_threshold(probs, labels, target_class, min_precision=0.90):
    candidates = np.sort(np.unique(probs))

    best_threshold = None
    best_coverage = 0

    for threshold in candidates:
        if target_class == 0:
            accepted = probs <= threshold
        else:
            accepted = probs >= threshold

        if accepted.sum() < 5:
            continue

        precision = np.mean(labels[accepted] == target_class)

        if precision >= min_precision and accepted.sum() > best_coverage:
            best_threshold = threshold
            best_coverage = accepted.sum()

    return best_threshold


real_threshold = find_threshold(
    val_probs, val_y, target_class=0
)

fake_threshold = find_threshold(
    val_probs, val_y, target_class=1
)

print("REAL threshold:", real_threshold)
print("FAKE threshold:", fake_threshold)
````
